```python
import os
import xml.etree.ElementTree as ET
from collections import defaultdict

import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms, models, utils
from PIL import Image

from sklearn.metrics import average_precision_score
import copy

# ===============================
# 1. Custom Dataset for VOC2008
# ===============================
class PascalVOCDataset(Dataset):
    def __init__(self, voc_root, split='trainval', transform=None):
        """
        voc_root: path to VOCdevkit/VOC2008 folder.
        split: split file name without extension (e.g. 'trainval' or 'test')
        transform: torchvision transforms to apply.
        """
        self.voc_root = voc_root
        self.transform = transform
        self.jpeg_dir = os.path.join(voc_root, 'JPEGImages')
        self.ann_dir  = os.path.join(voc_root, 'Annotations')
        self.imagesets_dir = os.path.join(voc_root, 'ImageSets', 'Main')

        # Read image IDs from the split file
        split_file = os.path.join(self.imagesets_dir, split + '.txt')
        with open(split_file, 'r') as f:
            self.image_ids = [line.strip() for line in f.readlines()]

        self.data = []  # list of tuples (img_path, label)
        for img_id in self.image_ids:
            img_path = os.path.join(self.jpeg_dir, img_id + '.jpg')
            ann_path = os.path.join(self.ann_dir, img_id + '.xml')
            if os.path.exists(ann_path):
                tree = ET.parse(ann_path)
                root = tree.getroot()
                objects = root.findall('object')
                if len(objects) > 0:
                    # Use first object’s name as class label.
                    label = objects[0].find('name').text
                    self.data.append((img_path, label))

        # Get unique sorted class labels and create a mapping.
        self.classes = sorted(list({label for (_, label) in self.data}))
        self.class_to_idx = {cls: i for i, cls in enumerate(self.classes)}
        print(f"Loaded {len(self.data)} images with {len(self.classes)} classes.")

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img_path, label = self.data[idx]
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        label_idx = self.class_to_idx[label]
        return image, label_idx

# ===============================
# 2. Define a Convolutional VAE
# ===============================
class ConvVAE(nn.Module):
    def __init__(self, latent_dim=128):
        super(ConvVAE, self).__init__()
        self.latent_dim = latent_dim
        # Encoder: assume input 3 x 64 x 64
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=4, stride=2, padding=1),   # 32 x 32 x 32
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=4, stride=2, padding=1),    # 64 x 16 x 16
            nn.ReLU(),
            nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1),   # 128 x 8 x 8
            nn.ReLU(),
            nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1),  # 256 x 4 x 4
            nn.ReLU()
        )
        self.fc_mu = nn.Linear(256*4*4, latent_dim)
        self.fc_logvar = nn.Linear(256*4*4, latent_dim)
        self.fc_decode = nn.Linear(latent_dim, 256*4*4)
        # Decoder: reconstruct to 3 x 64 x 64
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(256, 128, kernel_size=4, stride=2, padding=1),  # 128 x 8 x 8
            nn.ReLU(),
            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1),   # 64 x 16 x 16
            nn.ReLU(),
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1),    # 32 x 32 x 32
            nn.ReLU(),
            nn.ConvTranspose2d(32, 3, kernel_size=4, stride=2, padding=1),     # 3 x 64 x 64
            nn.Sigmoid()  # output in [0,1]
        )

    def encode(self, x):
        h = self.encoder(x)
        h = h.view(x.size(0), -1)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5*logvar)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        h = self.fc_decode(z)
        h = h.view(z.size(0), 256, 4, 4)
        x_recon = self.decoder(h)
        return x_recon

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        recon = self.decode(z)
        return recon, mu, logvar

def vae_loss(recon, x, mu, logvar):
    BCE = nn.functional.binary_cross_entropy(recon, x, reduction='sum')
    KLD = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return BCE + KLD

def train_vae(model, dataloader, device, epochs=20, lr=1e-3):
    optimizer = optim.Adam(model.parameters(), lr=lr)
    model.to(device)
    model.train()
    for epoch in range(1, epochs+1):
        running_loss = 0.0
        for images, _ in dataloader:
            images = images.to(device)
            optimizer.zero_grad()
            recon, mu, logvar = model(images)
            loss = vae_loss(recon, images, mu, logvar)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()
        avg_loss = running_loss / len(dataloader.dataset)
        print(f"VAE Epoch {epoch}/{epochs}: Loss = {avg_loss:.4f}")
    return model

def generate_vae_samples(model, device, num_samples=100):
    model.eval()
    with torch.no_grad():
        z = torch.randn(num_samples, model.latent_dim).to(device)
        samples = model.decode(z).cpu()
    return samples

# ===============================
# 3. Classifier Training with Transfer Learning
# ===============================
def get_pretrained_classifier(model_name, num_classes):
    if model_name.lower() == 'vgg16':
        model = models.vgg16(pretrained=True)
        # Replace classifier last layer
        in_features = model.classifier[6].in_features
        model.classifier[6] = nn.Linear(in_features, num_classes)
    elif model_name.lower() == 'resnet50':
        model = models.resnet50(pretrained=True)
        in_features = model.fc.in_features
        model.fc = nn.Linear(in_features, num_classes)
    elif model_name.lower() == 'densenet121':
        model = models.densenet121(pretrained=True)
        in_features = model.classifier.in_features
        model.classifier = nn.Linear(in_features, num_classes)
    else:
        raise ValueError("Model name not supported. Choose from VGG16, ResNet50, DenseNet121.")
    return model

def train_classifier(model, train_loader, val_loader, device, epochs=10, lr=1e-3):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    model.to(device)
    model.train()
    for epoch in range(1, epochs+1):
        running_loss = 0.0
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item()
        avg_loss = running_loss / len(train_loader.dataset)
        print(f"Classifier Epoch {epoch}/{epochs}: Loss = {avg_loss:.4f}")
    # After training, evaluate on val set and compute mAP
    model.eval()
    all_targets = []
    all_scores  = []
    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            outputs = model(images)
            scores = torch.softmax(outputs, dim=1).cpu().numpy()
            all_scores.append(scores)
            all_targets.append(labels.numpy())
    all_scores = np.concatenate(all_scores, axis=0)
    all_targets = np.concatenate(all_targets, axis=0)
    # One-hot encode ground truth
    n_samples = all_targets.shape[0]
    n_classes = all_scores.shape[1]
    targets_onehot = np.zeros((n_samples, n_classes))
    targets_onehot[np.arange(n_samples), all_targets] = 1
    # Compute average precision per class
    ap_list = []
    for c in range(n_classes):
        ap = average_precision_score(targets_onehot[:, c], all_scores[:, c])
        ap_list.append(ap)
    mAP = np.mean(ap_list)
    return mAP

# ===============================
# 4. Augment Training Data Function
# ===============================
def augment_training_data(original_dataset, vae_models, synth_counts, device):
    """
    For each class in vae_models, generate synth_counts synthetic samples.
    Then combine these with original_dataset.

    Returns a new dataset list of tuples (image_tensor, label).
    original_dataset: dataset instance from PascalVOCDataset.
    vae_models: dictionary mapping class_name to trained VAE model.
    synth_counts: number of synthetic samples to generate per class.
    """
    # Group original data by label index.
    class_to_data = defaultdict(list)
    for img, label in original_dataset:
        class_to_data[label].append((img, label))
    augmented_data = []

    # For each class, generate synthetic samples using corresponding VAE.
    # Note: We assume that vae_models keys are class names.
    for cls_name, model in vae_models.items():
        label_idx = original_dataset.class_to_idx[cls_name]
        # Add original images for this class.
        augmented_data.extend(class_to_data[label_idx])
        # Generate synthetic samples.
        synth_imgs = generate_vae_samples(model, device, num_samples=synth_counts)
        for img in synth_imgs:
            augmented_data.append((img, label_idx))
    return augmented_data

# ===============================
# 5. Main Experiment: Classification with and without Augmentation
# ===============================
if __name__ == '__main__':
    # Set paths and device.
    voc_root = '/content/VOCdevkit/VOC2008'  # Adjust path as needed.
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    # Define transforms.
    transform = transforms.Compose([
        transforms.Resize((64,64)),
        transforms.ToTensor()
    ])

    # Create the full dataset using 'trainval' for training.
    full_dataset = PascalVOCDataset(voc_root=voc_root, split='trainval', transform=transform)

    # For simplicity, we use a random 80/20 split from full_dataset for training and testing.
    num_total = len(full_dataset)
    indices = np.arange(num_total)
    np.random.shuffle(indices)
    split = int(np.floor(0.8 * num_total))
    train_indices, val_indices = indices[:split], indices[split:]

    train_dataset = Subset(full_dataset, train_indices)
    val_dataset   = Subset(full_dataset, val_indices)

    print(f"Training samples: {len(train_dataset)}, Validation samples: {len(val_dataset)}")

    # -------------------------------
    # 5a. Train per-class VAE on training data only.
    # -------------------------------
    # Group training data indices by class.
    class_to_indices = defaultdict(list)
    # full_dataset.data holds tuples (img_path, label) and full_dataset.class_to_idx maps class name -> index.
    # We need to map based on the subset indices.
    for sub_idx in train_dataset.indices:
        _, label = full_dataset.data[sub_idx]
        label_idx = full_dataset.class_to_idx[label]
        class_to_indices[label_idx].append(sub_idx)

    vae_models = {}
    batch_size = 32
    latent_dim = 128
    vae_epochs = 20
    vae_lr = 1e-3

    # For each class in the training split, train a VAE.
    for label_idx, indices in class_to_indices.items():
        class_name = full_dataset.classes[label_idx]
        print(f"\nTraining VAE for class '{class_name}' with {len(indices)} images.")
        subset = Subset(full_dataset, indices)
        loader = DataLoader(subset, batch_size=batch_size, shuffle=True)
        model = ConvVAE(latent_dim=latent_dim)
        model = train_vae(model, loader, device, epochs=vae_epochs, lr=vae_lr)
        vae_models[class_name] = model
        # (Optionally, show a grid of synthetic images for inspection.)
        synth = generate_vae_samples(model, device, num_samples=16)
        grid = utils.make_grid(synth, nrow=4, normalize=True)
        plt.figure(figsize=(6,6))
        plt.imshow(np.transpose(grid.numpy(), (1,2,0)))
        plt.title(f"Synthetic Samples for class '{class_name}'")
        plt.axis('off')
        plt.show()

    # -------------------------------
    # 5b. Transfer Learning: Evaluate three classifier architectures on original training set.
    # -------------------------------
    classifier_names = ['vgg16', 'resnet50', 'densenet121']
    num_classes = len(full_dataset.classes)

    # Define DataLoaders for classification.
    train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
    val_loader   = DataLoader(val_dataset, batch_size=32, shuffle=False)

    baseline_map = {}
    for name in classifier_names:
        print(f"\nTraining baseline classifier: {name.upper()}")
        model = get_pretrained_classifier(name, num_classes=num_classes)
        map_score = train_classifier(model, train_loader, val_loader, device, epochs=5, lr=1e-3)
        baseline_map[name] = map_score
        print(f"Baseline mAP for {name.upper()}: {map_score:.4f}")

    # -------------------------------
    # 5c. Classification using Augmented Data.
    # -------------------------------
    # For each augmentation level (100, 200, 500 synthetic samples per class)
    synth_levels = [100, 200, 500]
    augmented_results = {name: [] for name in classifier_names}

    for n_synth in synth_levels:
        print(f"\n--- Augmentation: Adding {n_synth} synthetic samples per class to training data ---")
        # Create an augmented training dataset (a list of (image, label) tuples)
        augmented_data = augment_training_data(full_dataset, vae_models, n_synth, device)

        # Create DataLoader over augmented data.
        class AugmentedDataset(Dataset):
            def __init__(self, data, transform=None):
                self.data = data
                self.transform = transform
            def __len__(self):
                return len(self.data)
            def __getitem__(self, idx):
                img, label = self.data[idx]
                if self.transform:
                    img = self.transform(img)
                return img, label

        aug_dataset = AugmentedDataset(augmented_data)
        aug_train_loader = DataLoader(aug_dataset, batch_size=32, shuffle=True)

        for name in classifier_names:
            print(f"\nTraining classifier {name.upper()} with augmented data (n_synth = {n_synth})")
            model = get_pretrained_classifier(name, num_classes=num_classes)
            # It is recommended to use the same hyperparameters.
            map_score = train_classifier(model, aug_train_loader, val_loader, device, epochs=5, lr=1e-3)
            augmented_results[name].append(map_score)
            print(f"mAP for {name.upper()} with {n_synth} augmentations per class: {map_score:.4f}")

    # -------------------------------
    # 5d. Plot the mAP curves for each classifier architecture.
    # -------------------------------
    plt.figure(figsize=(8,6))
    for name in classifier_names:
        plt.plot(synth_levels, augmented_results[name], marker='o', label=name.upper())
        # Also plot the baseline as a horizontal line.
        baseline = baseline_map[name]
        plt.hlines(baseline, synth_levels[0], synth_levels[-1], linestyles='dashed', colors='gray')
    plt.xlabel("Number of Synthetic Samples per Class")
    plt.ylabel("Mean Average Precision (mAP)")
    plt.title("Classifier Performance vs. Augmentation Level")
    plt.legend()
    plt.grid(True)
    plt.show()

    # -------------------------------
    # End of Experiment.
    # -------------------------------
    print("\nExperiment completed.")
```
